## 2. Chemical Engineering Context

Chemical engineers frequently deal with equations that describe how a quantity changes with time. For example, the concentration of a reactant in a batch reactor changes as a reaction proceeds.

Consider the simple first-order reaction:

```text
A → Products
```

For a first-order reaction, the rate of disappearance of A can be represented by:

```text
dC/dt = -kC
```

where:

- `C` = concentration of reactant A (mol/L)
- `t` = time (min)
- `k` = first-order rate constant (1/min)

Instead of solving the differential equation analytically, we will use **numerical integration**. This means that we calculate the concentration step-by-step over small time intervals.

**Note:** This is an educational numerical model for programming practice. It is not a substitute for a rigorous reactor-design calculation.


## 3. Euler Method – The Basic Idea

For a small time step `Δt`, the Euler method approximates the next concentration using:

```text
C_new = C_old + (dC/dt) × Δt
```

Since:

```text
dC/dt = -kC
```

we obtain:

```text
C_new = C_old - k × C_old × Δt
```

The important programming idea is that this calculation is repeated many times.

````python
# Activity 3
# Use a for loop to numerically integrate the first-order reaction for 10 minutes.
c=1.00
k=0.05
dt=1
t=10
for i in range(t+1):
    #print(f"Time: {i} minutes, Concentration: {c:.4f}")
    dCdt=-k * c
    c = c + dCdt * dt
    print(f"Time: {i} minutes, Concentration: {c:.4f}") 
print (f"final concentration: {c:.4f}")
````
