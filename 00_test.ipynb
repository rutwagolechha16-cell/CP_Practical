```python
#get marks of a student and print the grade
#above 90 A
#80-90 B
#70-80 C        
marks=float(input("Enter your marks: "))


if marks>90:
    print("Your grade is A")
elif marks>80:  
    print("Your grade is B")
elif marks>70:
    print("Your grade is C")
elif marks>60:
    print("Your grade is D")    
else:
    print("Your grade is F")

```

```python
# Updated grading system: 95 and above is A
marks = float(input("Enter your marks: "))

if marks >= 95:
    print("Your grade is A+")
elif marks >= 90:
    print("Your grade is A")        
elif marks >= 85:
    print("Your grade is B")
elif marks >= 75:
    print("Your grade is C")
elif marks >= 65:
    print("Your grade is D")
else:
    print("Your grade is F")
```
